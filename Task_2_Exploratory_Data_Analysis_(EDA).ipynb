```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
sns.set_theme(style="whitegrid")
```

```python
iris = load_iris(as_frame= True)
df = iris.frame
df["species"] = df["target"].map(dict(enumerate(iris.target_names)))
df = df.drop(columns=["target"])
print("Dataset loaded successfully!")
```

```python
display(df.head())
print("Dataset shape:",df.shape)
df.info()
display(df.describe())
print("Missing values:")
print(df.isnull().sum())
print("Duplicate rows:",df.duplicated().sum())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df,
             x= "sepal length (cm)",
             bins=15,
             kde=True)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length(cm)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
    data = df,
    x = "species",
    y = "petal length (cm)",
    )
plt.title("Average Petal Length by Species")
plt.xlabel("Flower Species")
plt.ylabel("Average Petal Length")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data = df,
    x = "petal length (cm)",
    y = "petal width (cm)",
    hue = "species"
)
plt.title("Relationship Between Petal Length and Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data = df,
    x = "species",
    y = "petal length (cm)"
)
plt.title("Box Plot of Petal Length by Species")
plt.xlabel("Flower Species")
plt.ylabel("Petal Length (cm)")
plt.tight_layout ()
plt.show()
```

##Dataset Overview
The Iris dataset contains flower measurements and species labels. I examined the dataset's dimensions, column names, data types, descriptive statistics, missing values, and duplicate records.

##Visualisations

Histogram: Used to examine the distribution of sepal length.

Bar chart: Used to compare average petal length across flower species.

##Key Findings

1. Dataset Overview:The Iris dataset contains 150 rows and 5 columns, including four numerical flower measurements and one categorical variable representing flower species.

2. Missing Values: No missing values were found in any of the five columns, indicating that the dataset is complete with respect to missing data.

3. Duplicate Records: One duplicate row was identified in the dataset. It was counted during the analysis but was not removed.

4. Sepal Length Distribution: The histogram shows the distribution of sepal length across the flowers. Sepal length ranges from 4.3 cm to 7.9 cm, with an average of approximately 5.84 cm.

5. Comparison of Flower Species: The bar chart shows differences in average petal length among the three species. Setosa has the shortest average petal length, followed by Versicolor, while Virginica has the longest.

6. Relationship Between Petal Measurements: The scatter plot indicates a positive relationship between petal length and petal width. Flowers with longer petals generally tend to have wider petals.

7. Species Separation: The scatter plot shows that Setosa is clearly separated from the other two species based on petal measurements. Versicolor and Virginica show some separation but may overlap.

8. Outlier Analysis: The box plot was used to examine petal length across the three species and identify potential outliers. Any potential outliers should be investigated before drawing conclusions about data errors.

## Conclusion
The exploratory data analysis provided a better understanding of the Iris dataset through descriptive statistics and visualisations. The analysis revealed differences in petal measurements among flower species and a positive relationship between petal length and width. It also identified one duplicate record and confirmed that no missing values were present.
