```python
import pandas as pd
data = """Student_ID,Name,Age,Gender,Marks,Department
101,Ali,20,Male,78,Economics
102,Sara,21,Female,85,Data Science
103,Rahul,,Male,72,Economics
104,Aisha,22,Female,88,Economics
105,John,20,Male,65,Data Science
105,John,20,Male,65,Data Science
106,Zoya,21,female,,data science
107,Ahmed,23,Male,75,Economics
108,Mehak,,Female,91,Data Science
109,Arjun,22,Female,69,Economics
"""

with open("Student_Data_Cleaning.csv", "w") as file:
    file.write(data)
```

```python
df = pd.read_csv("Student_Data_Cleaning.csv")
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df = df.drop_duplicates()
df['Gender'] = df['Gender'].str.title()
df['Department'] = df['Department'].str.title()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Marks'] = df['Marks'].fillna(df['Marks'].median())
```

```python
df['Age'].dtype
```

```python
df["Age"]
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Age"] = df["Age"].astype(int)
df.dtypes
print(df["Age"])
```

```python
print("Mean Age:", df["Age"].mean())
print("Median Age:", df["Age"].median())
```

```python
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
```

```python
print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
```

```python
print(df['Gender'].value_counts())
```

```python
print(df['Department'].value_counts())
```

#Data Cleaning & Exploration with Pandas

##Objective

To clean and explore a sample student dataset using the Pandas library in Python.

##Dataset Description

The dataset contains student information, including Student ID, Name, Age, Gender, Marks, and Department. Initially, it contained 10 rows and 6 columns.

##Cleaning Steps Performed

1. Loaded the CSV dataset using pd.read_csv().

2. Checked missing values using isnull().sum().

3. Identified and removed one duplicate row using drop_duplicates().

4. Handled missing Age and Marks values by replacing them with their respective medians.

5. Standardized Gender and Department text using str.title().

6. Converted Age and Marks to numeric data types using pd.to_numeric().

7. Converted Age to an integer data type.

##Data Exploration

Mean Age: approximately 21.22 years.

Median Age: 21 years.

Mean Marks: approximately 77.72.

Median Marks: 76.5.

Gender counts: Female – 5, Male – 4.

Department counts: Economics – 5, Data Science – 4.

##Findings

The original dataset contained one duplicate row, two missing Age values, and one missing Marks value. After cleaning, the dataset contained 9 rows and 6 columns, with no duplicate rows or missing values.
